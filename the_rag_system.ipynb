This improves the relevance of the evidence selected for generation; however, the scores do not represent fault probabilities or confirm a root cause.

```python
import hashlib
import pandas as pd
from pathlib import Path

from backend.rag_system.the_rag_pipeline import RAGPipeline

# Initialize existing RAG components
rag_pipeline = RAGPipeline()

retrieval_records = []

# Process each anomalous observation
for index, anomaly in enumerate(anomaly_records, start=1):

    print(f"Processing anomaly {index}/{len(anomaly_records)}")

    # Generate diagnostic query
    anomaly_details = rag_pipeline._build_anomaly_details(anomaly)

    transformed = rag_pipeline.query_transformer.transform(
        telemetry=anomaly,
        anomaly_details=anomaly_details
    )

    query = transformed["diagnostic_query"]
    asset_type = transformed["diagnostic_context"]["asset_type"]

    # Retrieve top 20 candidate documents
    retrieved_docs = rag_pipeline.retriever.retrieve(
        query=query,
        top_k=10,
        asset_type=asset_type
    )

    # Rerank and select top 5
    reranked_docs = rag_pipeline.reranker.rerank(
        query=query,
        documents=retrieved_docs,
        top_k=5
    )

    # Identify documents in the final top 5
    top5_ids = set()

    for doc in reranked_docs:
        content = (
            doc.get("page_content", "")
            if isinstance(doc, dict)
            else getattr(doc, "page_content", str(doc))
        )

        top5_ids.add(
            hashlib.sha256(content.encode("utf-8")).hexdigest()[:12]
        )

    # Store candidate documents
    for rank, doc in enumerate(retrieved_docs, start=1):

        if isinstance(doc, dict):
            content = doc.get("page_content", doc.get("text", ""))
            metadata = doc.get("metadata", {})
        else:
            content = getattr(doc, "page_content", str(doc))
            metadata = getattr(doc, "metadata", {})

        content = str(content).strip()

        if not content:
            continue

        doc_id = hashlib.sha256(
            content.encode("utf-8")
        ).hexdigest()[:12]

        retrieval_records.append({
            "asset_id": anomaly["asset_id"],
            "timestamp_id": anomaly["timestamp_id"],
            "diagnostic_query": query,
            "document_id": doc_id,
            "retrieval_rank": rank,
            "is_in_reranked_top5": doc_id in top5_ids,
            "document_content": content,
            "is_relevant": ""
        })


# Create evaluation DataFrame
retrieval_df = pd.DataFrame(retrieval_records)

print("\nRetrieval completed.")
print("Total candidate documents:", len(retrieval_df))

display(
    retrieval_df[
        [
            "asset_id",
            "document_id",
            "retrieval_rank",
            "is_in_reranked_top5",
            "document_content",
            "is_relevant"
        ]
    ]
)
```
